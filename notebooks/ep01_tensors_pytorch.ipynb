## Numpy and pytorch bridge

```python
array = np.array([1., 2, 3], dtype = np.float32)
shared = torch.from_numpy(array)
array[0] = 9.0
print("Tensor after numpy edit: ", shared)
```

```python
shared
```

```python
back = shared.numpy()
print(type(back))
```

```python
independent = shared.clone()

print(type(independent))
```

```python
back[1] = 8

print(shared)
print("independent copy: ", independent)
```

```python
independent.device
```

## Move tensors to the GPU

```python
torch.cuda.is_available()
```

```python
if torch.cuda.is_available():
    device = torch.device("cuda")
elif torch.backends.mps.is_available():
    device = torch.device("mps")
else:
    device = torch.device("cpu")

values = torch.tensor([1., 2,3]).to(device)
doubled = values * 2
print("Selected device: ", device)
print("Result on CPU: ", doubled.cpu())
```

## Autograd

```python
x = torch.tensor(3.0, requires_grad=True)
y = x**2

print("y: ", y.item())
```

```python
print("Gradient before: ", x.grad)
```

```python
y.backward()
```

```python
print("Gradient after: ", x.grad)
```

```python
print("change in X: ", x.item())
```

## Gradients accumulation

```python
new_y = x ** 2
new_y.backward()
print("Accumulated: ", x.grad.item())
x.grad = None

fresh_y = x ** 3

fresh_y.backward()
print("After clearing: ", x.grad.item())

```
